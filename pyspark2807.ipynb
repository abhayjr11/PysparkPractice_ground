```python
df=df.withColumn('revenue',col('quantity') * col('price') -col('shipping_cost')+col('discount'))
df.show(1)
```

```python
#Find the total revenue generated by each category.
df.groupBy('category').agg(sum('revenue')).show()

#Find the top 3 cities with the highest sales.
df.groupBy('city').agg(sum('revenue').alias('total_rev')).orderBy(desc('total_rev')).show(3)

#Calculate the average discount for each brand.
df.groupBy('brand').agg(round(avg('discount')).alias('avg_dis')).orderBy(desc('avg_dis')).show(1)

df.show(1)
```

```python
#Find the warehouse that shipped the maximum number of orders.
df.groupBy('warehouse').agg(count('product_name').alias('item_count')).orderBy(desc('item_count')).show()
#Calculate total shipping cost by payment method.
df.groupBy('payment_method').agg(sum('shipping_cost').alias('total_ship')).orderBy('total_ship').show()
#Find customers who placed more than one order.
df.groupBy('customer_id','customer_name').agg(count('order_id').alias('total_count')).filter(col('total_count')>1).show()
#Find categories having more than 5 products sold.
df.groupBy('category').agg(sum('quantity').alias('t_pro_sold')).filter(col('t_pro_sold')>5).show()
```

```python
#Calculate the average rating for each warehouse.
df.groupBy('warehouse').agg(round(avg('rating'))).show()
#Find the highest-priced product in each category.
win = Window.partitionBy('category').orderBy('price')
df.withColumn('rn', row_number().over(win)).filter(col('rn')==1).select('category','product_name','price').show()
df.show(1)
```

```python
#Calculate delivery time (days) for every order.
df=df.withColumn('days',datediff(col('delivery_date'),col('ship_date')))
#Find orders delivered in more than 5 days.
df.filter(col('days')>=5).show()
#Find monthly sales revenue.
df = df.withColumn('order_month',month('order_date'))
df.groupBy('order_month').agg(sum('revenue')).show()
#Find weekday-wise order count.
#Find monthly sales revenue.
df = df.withColumn('order_day',day('order_date'))
df.groupBy('order_day').agg(sum('revenue')).orderBy('order_day').show(1)
#Find customers who ordered in both January and February.
df.show(1)
```

```python
#Replace missing coupon codes with "NO_COUPON".
df = df.fillna({"coupon_code": "NO_COUPON"})
#Standardize customer names to Title Case.
df = df.withColumn('customer_name' ,initcap('customer_name'))
#Remove duplicate orders based on order_id.
df = df.dropDuplicates(['order_id'])
#Find cities with both delivered and returned orders.
df.select('city').filter((col('order_status')=='Delivered') | (col('order_status')=='Returned')).distinct().show()
df.show(1)
```

```python
from pyspark.sql.catalog import CatalogMetadata
#Rank products by price within each category.
win = Window.partitionBy('category').orderBy(desc('price'))
df.withColumn('rank',rank().over(win)).show(1)
#Find the second most expensive product in every category.
df.withColumn('rank',rank().over(win)).filter(col('rank')==2).show()
#Find the top 3 customers by spending.
win = Window.partitionBy('customer_id')
df.withColumn('s_sum',sum(col('quantity') * col('price')).over(win)).orderBy(desc('s_sum')).show(3)
#Find each customer's first order.
win = Window.partitionBy('customer_id').orderBy(desc('order_date'))
df.withColumn('rank',rank().over(win)).filter(col('rank')==1).show()
#Find each customer's latest order.
win = Window.partitionBy('customer_id').orderBy('order_date')
df.withColumn('rank',rank().over(win)).filter(col('rank')==1).show(1)
```

```python
#Calculate cumulative revenue by order date.
win = Window.orderBy('order_date').rowsBetween(Window.unboundedPreceding , Window.currentRow)
df.withColumn('c_sum', sum('revenue').over(win)).show(1)
#Calculate cumulative quantity sold by category.
win = Window.partitionBy('category').rowsBetween(Window.unboundedPreceding, Window.currentRow).orderBy('order_date')
df.withColumn('c_qunt', sum('quantity').over(win)).show(1)
#Find the previous order date for every customer.
win=Window.partitionBy('customer_id').orderBy('order_date')
df.withColumn('pre_date',lag('order_date').over(win)).show(1)
#Calculate the gap (days) between consecutive orders of a customer.
df.groupBy('customer_name').agg(count('order_id').alias('ord_c')).select('customer_name','ord_c').show(1)
#Calculate moving average of sales over the last 3 orders.

```

```python
#Find the customer contributing the highest percentage of total revenue.
win = Window.partitionBy('customer_id','customer_name')
t_rev = df.agg(sum('revenue')).first()[0]
df.withColumn('per_con', sum('revenue').over(win)).withColumn('contri', round(col('per_con')*100/t_rev,2)).show(1)
```
