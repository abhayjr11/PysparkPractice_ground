```python
pd.read_sql("select * from orders",conn).head(1)
```

```python
#Find orders from Delhi.

pd.read_sql("select * from orders where city == 'Delhi'",conn).head(1)
```

```python
#Find orders after 15 Jan.
pd.read_sql("select * from orders where order_date>'2024-01-15';", conn).head(1)
```

```python
# Show unique cities.
pd.read_sql("select distinct(city) from orders;", conn).head(1)
```

```python
#Count distinct customers.
pd.read_sql("select count(distinct(customer_id)) from orders;", conn)
```

```python
#Find total sales.
pd.read_sql("select sum(price*quantity) as tsales from orders;", conn)
```

```python
#Find average price.
pd.read_sql("select avg(price) as avg_prc from orders;", conn)
```

```python
#Find average quantity.
pd.read_sql("select avg(quantity) as avg_q from orders;", conn)
```

```python
#Find total revenue for Electronics.
pd.read_sql("select sum(price*quantity) as t_s_e from orders where category == 'Electronics';", conn)
```

```python
#Total sales by city.
pd.read_sql("select city, sum(price*quantity) as t_s_b_c from orders group by city;", conn)
```

```python
#Total revenue by customer.
pd.read_sql("select customer_id,sum(price*quantity) as r_b_c from orders group by customer_id;", conn).head(1)
```

```python
#Number of orders per city.
pd.read_sql("select city, count(customer_id) as t_c_b_c from orders group by city;", conn)
```

```python
#Average price by category.
pd.read_sql("select category, avg(price) as a_p_b_c from orders group by category;", conn)
```

```python
#Highest revenue city.
pd.read_sql("select city, sum(price*quantity) as max_rev from orders group by city order by max_rev desc limit 1;", conn)
```

```python
#Lowest revenue city.
pd.read_sql("select city, sum(price*quantity) as max_rev from orders group by city order by max_rev limit 1;", conn)
```

```python
#Product sold most.
pd.read_sql("select product, count(quantity) from orders group by product order by count(quantity) desc limit 1;", conn)
```

```python
#Category generating highest revenue.
pd.read_sql("select category, sum(price*quantity) as rev from orders group by category order by rev desc limit 1;", conn)
```

```python
#Orders by salesperson.
pd.read_sql("select salesperson, sum(quantity) as t_ord from orders group by salesperson ;", conn)
```

```python
#Revenue by payment mode.
pd.read_sql("select * from orders where order_date>'2024-01-15';", conn).head(1)
pd.read_sql("select payment_mode, sum(price*quantity)as t_rev from orders group by payment_mode;", conn)
```

```python
# Customers with more than one order.
pd.read_sql("select customer_id , sum(quantity) from orders group by customer_id having sum(quantity) >1 ;", conn).head(1)
```

```python
#Cities having revenue above ₹100000.
pd.read_sql("select city , sum(price*quantity) as t_rev from orders group by city having sum(price*quantity) > 100000 ;", conn)
```

```python
# Categories with more than five orders.
pd.read_sql("select category, sum(quantity) as t_qun from orders group by category having sum(quantity)>5  ;", conn)
```

```python
#Salespersons with revenue above ₹200000.
pd.read_sql("select salesperson, sum(quantity * price ) as t_rev from orders group by salesperson having sum(quantity * price )>200000  ;", conn)
```

```python
#Sort by customer then date.
pd.read_sql("select customer_id, order_date from orders order by customer_id , order_date;", conn).head(1)
```

```python
#Uppercase customer names.
pd.read_sql("select upper(customer_name) as customer_name from orders;", conn).head(1)
```

```python
# Lowercase city names.
pd.read_sql("select lower(city) as city from orders;", conn).head(1)
```

```python
#Length of product names.
pd.read_sql("select product, length(product) as pro_len from orders group by product;", conn).head(1)
```

```python
#First three letters of customer names.
pd.read_sql("select upper(substr(customer_name,1,3)) as abb from orders ;", conn).head(1)
```

```python
# Replace Laptop with Notebook
pd.read_sql("select product, replace(product , 'Laptop' , 'Notebook') as updated_pro from orders ;", conn).head(1)
```

```python
# Concatenate customer and city.
pd.read_sql("select customer_name , city , customer_name || ' ' || city as cus_cit from orders ;", conn).head(1)
```

```python
#Customers whose name starts with J.
pd.read_sql("select customer_name from orders where customer_name like'j%';", conn)
```

```python
#Customers ending with a.
pd.read_sql("select customer_name from orders where customer_name like '%a' ;", conn).head(1)
```

```python
#Product containing "top".
pd.read_sql("select product from orders where product like '%top%' ;", conn).head(1)
```

```python
#Reverse customer names.
conn.create_function("reve",1,lambda x : x[::-1] if x else None )
pd.read_sql("select customer_name, reve(customer_name) as rev  from orders ;", conn).head(1)
```
